```python
env = gym.make("CartPole-v1")
total_reward = 0.0
total_steps = 0
obs = env.reset()
```

```python
while True:
    action = env.action_space.sample()
    obs, reward, _, done, _ = env.step(action)
    total_reward += float(reward)
    total_steps += 1
    if done:
        break
    if total_reward > 100:
        break
    print("Episode done in %d steps, total reward %.2f" % (
        total_steps, total_reward))
```
